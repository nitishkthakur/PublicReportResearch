# Lite Gramework RAG Example

```python
from langchain_agents import RAG
DOCS_DIR = 'docs'
INDEX_DIR = 'rag_index'
rag = RAG()
rag.build_vector_store(DOCS_DIR)
rag.save_vector_store(INDEX_DIR)
```

```python
rag2 = RAG()
rag2.load_vector_store(INDEX_DIR)
result = rag2.invoke('Provide a short summary of the earnings of citi bank in Q4 2024 and format it as a table. After the table, provide a sno')
print(result)
```
